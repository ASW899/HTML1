```python
from datetime import datetime


def display_leap_years(final_year):
  current_year = datetime.now().year
  print(f"\nLeap years from {current_year} to {final_year}:")

  for year in range(current_year, final_year + 1):
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
      print(year)


current_year = datetime.now().year


while True:
  user_final_year = int(input("Enter the final year: "))

  if user_final_year <= current_year:
    print("Enter a year greater than the current year.\n")
  else:
    display_leap_years(user_final_year)
    break  
```

```python
class Time:
    def set_time(self, hour, minute, second):
        self.__hour = hour
        self.__minute = minute
        self.__second = second

    def __add__(self, other):
        total_seconds = self.__second + other.__second
        total_minutes = self.__minute + other.__minute + (total_seconds // 60)
        total_hours = self.__hour + other.__hour + (total_minutes // 60)
        
        
        final_seconds = total_seconds % 60
        final_minutes = total_minutes % 60
        
        
        result = Time()
        result.set_time(total_hours, final_minutes, final_seconds)
        return result

    def display(self):
        print(f"Total Time -> {self.__hour:02d}:{self.__minute:02d}:{self.__second:02d}")


t1 = Time()
t2 = Time()

t1.set_time(5, 45, 40) 
t2.set_time(2, 25, 30)  


t3 = t1 + t2
t3.display()  
```
