```python
number = int(input("enter the number to check if the number is a disarium number "))
digits = str(number)
total = 0
for i in range(len(digits)):
    digit = int(digits[i])  
    position = i + 1        
    total = total + (digit ** position) 
if total == number:
    print(number, "is a Disarium number!")
else:
    print(number, "is not     a Disarium number.")
```
