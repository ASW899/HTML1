```python
numbers=[]
print("enter the numbers in the list")
for i in range(5):
    user_add=int(input(f" NUMBER{i+1}: "))
    numbers.append(user_add)
for num in numbers:
    if num%2==0:
        numbers.remove(num)
print("The list after removing the even numbers is: ",numbers)
```
